```python
import pandas as pd
```

```python
archivo = "/content/notas.csv"
notas = pd.read_csv(archivo, sep=";", encoding="utf-8")
print("Hola ", notas["nombres"][0], "te sacaste", notas["notas"][0], "en el lab.")
```

```python
aprobados = notas[notas["notas"] >= 6]
print(aprobados)
```

```python

reprobados = notas[notas["notas"] < 6]
print(reprobados)
```

```python
print( "Cantidad de alumnos aprobados", len(aprobados) )
print( "Cantidad de alumnos reprobados", len(reprobados) )
```

```python
cantidad_aprobados = notas.loc[notas["notas"] >= 6, "notas"].count()
cantidad_reprobados = notas.loc[notas["notas"] < 6, "notas"].count()

print("Cantidad de alumnos aprobados", cantidad_aprobados)
print("Cantidad de alumnos reprobados", cantidad_reprobados)
```

```python
cantidad_aprobados = notas.loc[notas["notas"] >= 6].agg({"notas": "count"})
cantidad_reprobados = notas.loc[notas["notas"] < 6].agg({"notas": "count"})

print("Cantidad de alumnos aprobados", cantidad_aprobados)
print("Cantidad de alumnos reprobados", cantidad_reprobados)
#
```

```python
notas_alumnos = notas.groupby(["notas"]).agg({'nombres': lambda n: ', '.join(map(str, n))})
print(notas_alumnos)
```

```python
#importar la libreria oara graficos
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
notas['estado'] = notas['notas'].apply(lambda x: 'Aprobado' if x >= 6 else 'Reprobado')
notas
```

```python
sns.set(style="whitegrid")
sns.countplot(
    x="estado",
    data=notas,
    hue="estado",
    palette={'Aprobado': 'green', 'Reprobado':'red'}
)
plt.title("Distribución de Notas")
plt.xlabel("Estado")
plt.ylabel("Cantidad")
plt.show()
```

```python
# Contar la cantidad de alumnos por estado
conteo = notas['estado'].value_counts()

# Crear la gráfica de pastel
plt.figure(figsize=(6, 6))
plt.pie(
    conteo,
    labels=conteo.index,
    autopct='%1.1f%%',
    startangle=140,
    colors=['green', 'red']
)
plt.title("Distribución de Notas (Aprobados vs Reprobados)")
plt.show()
```
