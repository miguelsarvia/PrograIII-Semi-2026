```python
#librerias a importar
import tensorflow as tf
import pandas as pd
import numpy as np
```

```python
#leer achivo
datos = pd.read_csv('/content/conversor.csv', sep=';', encoding='utf-8')
print(datos['CELCIUS'])
```

```python
#obtener datos
c = datos['CELCIUS']
f = datos['FARENHEITS']
print(c, f)
```

```python
#creamos el modelo
modelo = tf.keras.Sequential()
modelo.add(tf.keras.layers.Dense(units=1, input_shape=[1]))
```

```python
#compilar el modelo
modelo.compile(optimizer=tf.keras.optimizers.Adam(1.0), loss='mean_squared_error')
```

```python
#entrenar modelo
epocas = modelo.fit(c,f, epochs=200, verbose=1)
```

```python
respuesta= modelo.predict(np.array([100.0]))
print(respuesta)
```
