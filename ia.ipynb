### Paso 1: Agregar la columna Kelvin y preparar los datos de entrenamiento

```python
# Agregar la columna Kelvin matemáticamente a nuestro DataFrame
datos['KELVIN'] = datos['CELCIUS'] + 273.15

# Definir la entrada (Celsius)
c_input = datos['CELCIUS'].values

# Definir las salidas (Fahrenheit y Kelvin juntas)
targets = datos[['FARENHEITS', 'KELVIN']].values

print("Entrada (Celsius):\n", c_input[:5])
print("\nSalidas deseadas (Fahrenheit, Kelvin):\n", targets[:5])
```

```python
# 1. Importar librerías necesarias
import tensorflow as tf
import pandas as pd
import numpy as np
import os

# 2. Verificar y cargar el archivo original
archivo_ruta = '/content/conversor.csv'
if not os.path.exists(archivo_ruta):
    print("❌ ERROR: No se encontró el archivo 'conversor.csv' en la carpeta /content/")
    print("👉 PASO A SEGUIR: Haz clic en el icono de la carpeta (📁) a la izquierda de la pantalla, sube tu archivo 'conversor.csv' allí, y vuelve a ejecutar esta celda.")
else:
    datos = pd.read_csv(archivo_ruta, sep=';', encoding='utf-8')
    print("✅ ¡Archivo conversor.csv cargado correctamente!")

    # 3. Agregar columna Kelvin y preparar datos
    datos['KELVIN'] = datos['CELCIUS'] + 273.15
    c_input = datos['CELCIUS'].values
    targets = datos[['FARENHEITS', 'KELVIN']].values

    # 4. Guardar el nuevo CSV para tu entrega
    datos.to_csv('/content/conversor_completo.csv', sep=';', index=False, encoding='utf-8')
    print("✅ ¡Nuevo archivo 'conversor_completo.csv' generado con éxito en el panel de la izquierda!")

    # 5. Crear el modelo de doble salida
    modelo_doble = tf.keras.Sequential([
        tf.keras.layers.Input(shape=(1,)),
        tf.keras.layers.Dense(units=2)
    ])

    modelo_doble.compile(
        optimizer=tf.keras.optimizers.Adam(0.5),
        loss='mean_squared_error'
    )

    # 6. Entrenar la IA
    print("Entrenando el modelo, por favor espera...")
    modelo_doble.fit(c_input, targets, epochs=350, verbose=0)
    print("¡IA entrenada con éxito para convertir a Fahrenheit y Kelvin!")

    # 7. Demostración de predicción
    pred = modelo_doble.predict(np.array([[100.0]]))
    print(f"\nPrueba de predicción para 100 °C:")
    print(f"- Fahrenheit predicho: {pred[0][0]:.2f} °F (Esperado: 212.00)")
    print(f"- Kelvin predicho: {pred[0][1]:.2f} K (Esperado: 373.15)")
```

### Guardar los nuevos datos en un archivo CSV

```python
# Guardar el DataFrame modificado en un nuevo archivo CSV
datos.to_csv('/content/conversor_completo.csv', sep=';', index=False, encoding='utf-8')
print("¡Archivo 'conversor_completo.csv' guardado con éxito!")
```

### Paso 2: Crear y compilar el modelo con 2 neuronas de salida

```python
# Creamos el modelo exactamente igual de simple, pero con 2 salidas
modelo_doble = tf.keras.Sequential()
modelo_doble.add(tf.keras.layers.Dense(units=2, input_shape=[1]))

# Compilar el modelo
modelo_doble.compile(optimizer=tf.keras.optimizers.Adam(1.0), loss='mean_squared_error')
```

### Paso 3: Entrenar el modelo

```python
# Entrenar el modelo con más épocas para que asimile bien los Kelvin
epocas_doble = modelo_doble.fit(c_input, targets, epochs=1000, verbose=0)
print("¡Modelo doble entrenado!")
```

### Paso 4: Realizar una predicción

```python
# Realizar una predicción de prueba dentro del rango de tus datos (por ejemplo, 20°C)
valor_prueba = 20.0
prediccion = modelo_doble.predict(np.array([valor_prueba]))

print(f"Predicción para {valor_prueba}°C:")
print(f"- Fahrenheit predicho: {prediccion[0][0]:.2f} °F (Esperado: 68.00)")
print(f"- Kelvin predicho: {prediccion[0][1]:.2f} K (Esperado: 293.15)")
```
