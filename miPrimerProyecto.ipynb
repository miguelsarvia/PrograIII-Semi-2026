**Calcular el sueldo de un empleado. hrs * valor de horas * 30 dias las deducciones: ISSS, AFP, ISR (Ver tabla del MH)**

```python
def calcular_isr(sueldo):
    if sueldo <= 472.00:
        return 0
    elif sueldo <= 895.24:
        return (sueldo - 472.00) * 0.10 + 17.67
    elif sueldo <= 2038.10:
        return (sueldo - 895.24) * 0.20 + 60.00
    else:
        return (sueldo - 2038.10) * 0.30 + 288.57
```

```python
horas = float(input("Horas trabajadas por día: "))
valor_hora = float(input("Valor por hora: $"))

sueldo = horas * valor_hora * 30

isss = sueldo * 0.03
afp = sueldo * 0.0625
isr = calcular_isr(sueldo)

neto = sueldo - isss - afp - isr
print("\n========== BOLETA DE PAGO ==========")
print(f"Sueldo bruto : ${sueldo:.2f}")
print(f"ISSS (3%)    : ${isss:.2f}")
print(f"AFP (6.25%)  : ${afp:.2f}")
print(f"ISR          : ${isr:.2f}")
print("------------------------------------")
print(f"Sueldo neto  : ${neto:.2f}")
```

**Ejercicio encontrar la nota final de cada materia asiendo uso de funciones.**

```python
def calcular_nota_computo(computo_notas):
    """Calcula la nota de un cómputo (c1, c2, o c3) usando pesos 0.3, 0.3, 0.4."""
    lab1 = computo_notas["lab1"]
    lab2 = computo_notas["lab2"]
    parcial = computo_notas["parcial"]
    return (lab1 * 0.3) + (lab2 * 0.3) + (parcial * 0.4)

def calcular_nota_final_materia(materia):
    """Calcula la nota final de una materia promediando las notas de c1, c2 y c3."""
    notas_c1 = materia["notas"]["c1"]
    notas_c2 = materia["notas"]["c2"]
    notas_c3 = materia["notas"]["c3"]

    nota_c1 = calcular_nota_computo(notas_c1)
    nota_c2 = calcular_nota_computo(notas_c2)
    nota_c3 = calcular_nota_computo(notas_c3)

    return (nota_c1 + nota_c2 + nota_c3) / 3

print(f"Notas finales de {persona['nombre']}:")

# Extraer los nombres de las materias
nombres_materias = [materia["nombre"] for materia in persona["materias"]]

# Calcular las notas finales usando map
notas_finales = map(calcular_nota_final_materia, persona["materias"])

# Imprimir los resultados
for nombre, nota in zip(nombres_materias, notas_finales):
    print(f"- {nombre}: {nota:.2f}")
```

1. **Calcular** sobre el sueldo las deducciones por ley ISSS (3%), AFP(6.25%), ISR(Según tabla) e imprima el total deducciones y sueldo neto.

```python
while True:
    sueldo = float(input("Ingrese el sueldo: $"))

    # Cálculo de ISSS (3% con tope de $30.00)
    if sueldo * 0.03 > 30.00:
        isss = 30.00
    else:
        isss = sueldo * 0.03

    # Cálculo de AFP (7.25% con tope de $507.50)
    if sueldo * 0.0725 > 507.50:
        afp = 507.50
    else:
        afp = sueldo * 0.0725

    # Calcular la Renta Gravada
    renta_gravada = sueldo - isss - afp

    # Cálculo de ISR usando if/elif/else
    if renta_gravada <= 550.00:
        isr = 0.00
    elif renta_gravada <= 895.24:
        isr = (17.67 + (renta_gravada - 550.00) * 0.10)
    elif renta_gravada <= 2038.10:
        isr = (60.00 + (renta_gravada - 895.24) * 0.20)
    else:
        isr = (288.57 + (renta_gravada - 2038.10) * 0.30)

    total_deducciones = isss + afp + isr

    # Sueldo neto
    sueldo_neto = sueldo - total_deducciones

    print("\n--- DEDUCCIONES ---")
    print(f"Sueldo:             ${sueldo:.2f}")
    print(f"ISSS (3%):          ${isss:.2f}")
    print(f"AFP (7.25%):        ${afp:.2f}")
    print(f"ISR:                ${isr:.2f}")
    print(f"Total deducciones:  ${total_deducciones:.2f}")
    print(f"Sueldo neto:        ${sueldo_neto:.2f}")

    continuar = input("\n¿Desea calcular otro sueldo? (s/n): ").lower()
    if continuar != 's':
        break
print("Fin del cálculo de sueldo.")
```
