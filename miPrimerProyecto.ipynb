**Funciones en PYTHON**

```python
def suma(num1, num2):
  respuesta = float(num1) + float(num2)
  return respuesta
```

```python
def resta (num1, num2):
  respuesta = float(num1) - float(num2)
  return respuesta
```

```python
num1 = input("Num 1: ")
num2 = input("Num 2: ")
print("La suma: ", suma(num1, num2))
print("La Resta: ", resta(num1, num2))
```

**convertir grados fahrenheit a celsius**

```python
def convertir(fahrenheit):
  return (float(fahrenheit) - 32) * 5/9
```

```python
c = input("Grados fahrenheit:")
print ("Grados celsius:", convertir (c))
```

**convertir grados celsius a fahrenheit**

```python
def convertir(celsius):
  return float(celsius)*9/5 + 32
```

```python
c = input("Grados celsius:")
print ("Grados fafahrenheit:", convertir (c))
```

**Convertir de grados celsius a kelvin**

```python
def convertir(celsius):
  return float(celsius)+273.15
```

```python
c = input("Grados celsius:")
print ("Grados kelvin:", convertir (c))
```
