**Funciones en PYTHON**

```python
def suma(num1, num2):
  respuesta = float(num1) + float(num2)
  return respuesta
```

```python
def resta (num1, num2):
  respuesta = float(num1) - float(num2)
  return respuesta
```

**convertir grados fahrenheit a celsius**

```python
def convertir(fahrenheit):
  return (float(fahrenheit) - 32) * 5/9
```

```python
c = input("Grados fahrenheit:")
print ("Grados celsius:", convertir (c))
```

**Calcular el sueldo de un empleado. hrs * valor de horas * 30 dias las deducciones: ISSS, AFP, ISR (Ver tabla del MH)**

```python
def calcular_isr(sueldo):
    if sueldo <= 472.00:
        return 0
    elif sueldo <= 895.24:
        return (sueldo - 472.00) * 0.10 + 17.67
    elif sueldo <= 2038.10:
        return (sueldo - 895.24) * 0.20 + 60.00
    else:
        return (sueldo - 2038.10) * 0.30 + 288.57
```

```python
horas = float(input("Horas trabajadas por día: "))
valor_hora = float(input("Valor por hora: $"))

sueldo = horas * valor_hora * 30

isss = sueldo * 0.03
afp = sueldo * 0.0725
isr = calcular_isr(sueldo)

neto = sueldo - isss - afp - isr
print("\n========== BOLETA DE PAGO ==========")
print(f"Sueldo bruto : ${sueldo:.2f}")
print(f"ISSS (3%)    : ${isss:.2f}")
print(f"AFP (7.25%)  : ${afp:.2f}")
print(f"ISR          : ${isr:.2f}")
print("------------------------------------")
print(f"Sueldo neto  : ${neto:.2f}")
```
